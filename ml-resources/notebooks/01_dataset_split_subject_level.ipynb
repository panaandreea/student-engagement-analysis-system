```python
import random
```

```python
# group 01

random.seed(42)

right_subjects = ["S01", "S02", "S03", "S11", "S12", "S13"]
central_subjects = ["S04", "S05", "S06", "S07", "S14", "S15", "S16", "S17"]
left_subjects = ["S08", "S09", "S10", "S18", "S19", "S20"]

zones = {
    "right": right_subjects,
    "central": central_subjects,
    "left": left_subjects
}

split_config = {
    "right": {"train":4, "val":1, "test":1},
    "central": {"train":6, "val":1, "test":1},
    "left": {"train":4, "val":1, "test":1}
}

train, val, test = [], [], []

for zone, subjects in zones.items():
    subjects = subjects.copy()
    random.shuffle(subjects)

    n_test = split_config[zone]["test"]
    n_val = split_config[zone]["val"]

    test.extend(subjects[:n_test])
    val.extend(subjects[n_test:n_test + n_val])
    train.extend(subjects[n_test + n_val:])

print("TRAIN:", train)
print("VAL:", val)
print("TEST:", test)
```
